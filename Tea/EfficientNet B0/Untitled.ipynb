```python

import random
import shutil
from pathlib import Path

# =====================================
# SOURCE AND OUTPUT DIRECTORIES
# =====================================

SOURCE_DIR = Path(r"D:\Tea\Gray_Blight_Unique")
OUTPUT_DIR = Path(r"D:\Tea\Gray Blight")

# Split ratios
TRAIN_RATIO = 0.80
VAL_RATIO = 0.10
TEST_RATIO = 0.10

random.seed(42)

IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

# =====================================
# CHECK SOURCE DIRECTORY
# =====================================

if not SOURCE_DIR.exists():
    raise FileNotFoundError(f"Source directory not found: {SOURCE_DIR}")

# Find images, including images in subfolders
images = [
    file for file in SOURCE_DIR.rglob("*")
    if file.is_file() and file.suffix.lower() in IMAGE_EXTENSIONS
]

print("Total images found:", len(images))

if len(images) == 0:
    raise ValueError("No images found. Check the source directory and image formats.")

# =====================================
# SHUFFLE AND SPLIT
# =====================================

random.shuffle(images)

total = len(images)

train_end = int(total * TRAIN_RATIO)
val_end = train_end + int(total * VAL_RATIO)

train_images = images[:train_end]
val_images = images[train_end:val_end]
test_images = images[val_end:]

# =====================================
# CREATE OUTPUT FOLDERS
# =====================================

for split in ["train", "val", "test"]:
    (OUTPUT_DIR / split / "Gray_Blight").mkdir(
        parents=True, exist_ok=True
    )

# =====================================
# COPY IMAGES
# =====================================

def copy_images(image_list, split_name):
    destination = OUTPUT_DIR / split_name / "Gray_Blight"

    for index, image in enumerate(image_list, start=1):
        # Preserve original filename
        shutil.copy2(image, destination / image.name)

    print(f"{split_name}: {len(image_list)} images copied.")

copy_images(train_images, "train")
copy_images(val_images, "val")
copy_images(test_images, "test")

# =====================================
# FINAL SUMMARY
# =====================================

print("\n========== DATASET SPLIT SUMMARY ==========")
print("Total images:", total)
print("Training images:", len(train_images))
print("Validation images:", len(val_images))
print("Testing images:", len(test_images))
print("Output directory:", OUTPUT_DIR)
print("\nDataset splitting completed successfully!")
```
